### La solución es de tipo Binomial, porque se analiza un número fijo de ensayos independientes (n = 1000 motores), cada uno con dos posibles resultados (defectuoso o no) y con la misma probabilidad de ser defectuoso (p = 0.001). La variable aleatoria  X  = número de motores defectuosos en el lote sigue exactamente una distribución Binomial( n=1000, p=0.001 ).

```python
from scipy.stats import binom

def main():
    n = 1000      # tamaño del lote
    p = 0.001     # probabilidad de que un motor sea defectuoso

    # a) Probabilidad de que ninguno sea defectuoso: P(X = 0)
    prob_0 = binom.pmf(0, n, p)
    print(f"Probabilidad de que ningún motor sea defectuoso: {prob_0:.6f}")

    # b) Probabilidad de que exactamente tres sean defectuosos: P(X = 3)
    prob_3 = binom.pmf(3, n, p)
    print(f"Probabilidad de que exactamente 3 motores sean defectuosos: {prob_3:.6f}")

    # c) Probabilidad de que exactamente cuatro sean defectuosos: P(X = 4)
    prob_4 = binom.pmf(4, n, p)
    print(f"Probabilidad de que exactamente 4 motores sean defectuosos: {prob_4:.6f}")

if __name__ == "__main__":
    main()
```
