```python
from scipy.stats import binom

#Parámetros de La distribución binomial
n = 10
p = 0.10
x = 2

#Calcular La probilidad de la variable aleatoria X = 2
probabilidad_x_2 = binom.pmf(x, n, p)

print("La probabilidad de que la variale aleatoria se X se aigual a 2 es: ", probabilidad_x_2)
```
