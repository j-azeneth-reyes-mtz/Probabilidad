```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, linregress
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Configuración de gráficos
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Lista de grupos
grupos = ['grupo1.csv', 'grupo2.csv', 'grupo3.csv']
nombres = ['Grupo 1', 'Grupo 2', 'Grupo 3']

# Función para análisis completo de cada grupo
def analizar_grupo(archivo, nombre_grupo):
    print(f"\n{'='*80}")
    print(f"ANÁLISIS PARA {nombre_grupo.upper()} - {archivo}")
    print('='*80)

    # 1. Carga y exploración de datos
    df = pd.read_csv(archivo)
    print("\n--- Primeras filas del dataset ---")
    print(df.head())

    print("\n--- Descripción estadística ---")
    print(df.describe())

    # Histograma de calificaciones
    plt.figure()
    plt.hist(df['Calificacion'], bins=10, edgecolor='black', alpha=0.7, color='skyblue')
    plt.xlabel('Calificación Final')
    plt.ylabel('Frecuencia')
    plt.title(f'Histograma de Calificaciones - {nombre_grupo}')
    plt.grid(axis='y', alpha=0.75)
    plt.show()

    # 2. Coeficiente de correlación de Pearson
    horas = df['HorasEstudio']
    calif = df['Calificacion']
    corr, p_valor_corr = pearsonr(horas, calif)
    print(f"\n--- Coeficiente de correlación de Pearson ---")
    print(f"Coeficiente de correlación (r): {corr:.4f}")
    print(f"p-valor: {p_valor_corr:.4e}")

    # 3. Modelo de regresión lineal (statsmodels para significancia)
    # Añadir constante para el intercepto
    X = sm.add_constant(horas)
    modelo = sm.OLS(calif, X).fit()

    print("\n--- Modelo de Regresión Lineal (statsmodels) ---")
    print(modelo.summary())

    pendiente = modelo.params['HorasEstudio']
    intercepto = modelo.params['const']
    r_cuadrado = modelo.rsquared
    p_valor_reg = modelo.f_pvalue

    print(f"\nEcuación de la línea de regresión:")
    print(f"Calificación = {intercepto:.4f} + {pendiente:.4f} * Horas de estudio")

    # 4. Visualización: scatter plot + línea de regresión
    plt.figure()
    plt.scatter(horas, calif, alpha=0.7, color='blue', label='Datos reales')

    # Línea de regresión
    x_range = np.linspace(horas.min(), horas.max(), 100)
    y_pred = intercepto + pendiente * x_range
    plt.plot(x_range, y_pred, color='red', linewidth=2, label='Línea de regresión')

    plt.xlabel('Horas de estudio')
    plt.ylabel('Calificación Final')
    plt.title(f'Regresión Lineal - {nombre_grupo}\n$r = {corr:.3f}$, $R^2 = {r_cuadrado:.3f}$')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

    # 5. Interpretación de resultados
    print("\n--- INTERPRETACIÓN DE RESULTADOS ---")

    # Pregunta 1
    print(f"\n1. Interpretación del coeficiente de regresión (pendiente = {pendiente:.4f}):")
    print(f"   Por cada hora adicional de estudio, la calificación final aumenta en promedio {pendiente:.4f} puntos.")
    if pendiente > 0:
        print("   → Relación positiva: más estudio → mayor calificación.")
    else:
        print("   → Relación negativa: más estudio → menor calificación.")

    # Pregunta 2
    print(f"\n2. Interpretación del coeficiente de correlación (r = {corr:.4f}):")
    fuerza = ""
    if abs(corr) >= 0.8:
        fuerza = "muy fuerte"
    elif abs(corr) >= 0.6:
        fuerza = "fuerte"
    elif abs(corr) >= 0.4:
        fuerza = "moderada"
    elif abs(corr) >= 0.2:
        fuerza = "débil"
    else:
        fuerza = "muy débil"
    direccion = "positiva" if corr > 0 else "negativa"
    print(f"   → Correlación {direccion} y {fuerza} (r = {corr:.4f}).")
    print(f"   Indica una asociación lineal {direccion} entre horas de estudio y calificaciones.")

    # Pregunta 3
    print(f"\n3. ¿El modelo de regresión lineal es estadísticamente significativo?")
    print(f"   p-valor del modelo (F-test) = {p_valor_reg:.4e}")
    if p_valor_reg < 0.05:
        print("   → SÍ, el modelo es estadísticamente significativo (p < 0.05).")
        print("   La relación lineal observada no se debe al azar.")
    else:
        print("   → NO, el modelo NO es estadísticamente significativo (p >= 0.05).")
        print("   No hay suficiente evidencia para afirmar una relación lineal.")

    # Pregunta 4
    print(f"\n4. ¿Qué tan confiable es el modelo para predecir calificaciones?")
    print(f"   Coeficiente de determinación R² = {r_cuadrado:.4f}")
    if r_cuadrado >= 0.7:
        conf = "alta"
    elif r_cuadrado >= 0.5:
        conf = "moderada"
    else:
        conf = "baja"
    print(f"   → El modelo explica el {r_cuadrado*100:.2f}% de la variabilidad en las calificaciones.")
    print(f"   → Confiabilidad predictiva: {conf}.")
    if r_cuadrado < 0.5:
        print("   Otros factores no considerados tienen mayor influencia en el rendimiento.")

    return {
        'grupo': nombre_grupo,
        'correlacion': corr,
        'p_corr': p_valor_corr,
        'pendiente': pendiente,
        'intercepto': intercepto,
        'r2': r_cuadrado,
        'p_reg': p_valor_reg
    }

# Ejecutar análisis para los tres grupos
resultados = []
for archivo, nombre in zip(grupos, nombres):
    res = analizar_grupo(archivo, nombre)
    resultados.append(res)

# Resumen final comparativo
print("\n\n" + "="*80)
print("RESUMEN COMPARATIVO DE LOS TRES GRUPOS")
print("="*80)
resumen_df = pd.DataFrame(resultados)
print(resumen_df.round(4))

print("\n--- CONCLUSIONES GENERALES ---")
for i, res in enumerate(resultados):
    print(f"\n{res['grupo']}:")
    print(f"  r = {res['correlacion']:.3f}, R² = {res['r2']:.3f}, pendiente = {res['pendiente']:.3f}")
    if res['p_reg'] < 0.05:
        print("Modelo significativo")
    else:
        print("Modelo NO significativo")

print("\nNota: La existencia de una relación lineal depende del grupo.")
print("Si algún grupo muestra correlación baja y p-valor alto, no hay evidencia de relación lineal.")
```
